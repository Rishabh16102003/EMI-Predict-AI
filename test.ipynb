### fixing age Column

```python
import re
df['age'] = df['age'].astype(str)
df['age'] = df['age'].str.replace(r'\.0+', '', regex=True)
df['age'] = pd.to_numeric(df['age'], errors='coerce')
df['age'] = df['age'].astype(int)
```

```python
df['gender'].value_counts()
```

### Fixing Gender column

```python
gender_dict={
    'Male':'M',
    'MALE':'M',
    'male':'M',
    'Female':'F',
    'female':'F',
    'FEMALE':'F'
}
df['gender']=df['gender'].replace(gender_dict)
```

```python
df['education'].value_counts()
```

```python
df.head()
```

### fixing education column filling na with unknown

```python
df['education']=df['education'].fillna('Unknown')
```

### Fixing Column monthly_salary

```python
df['monthly_salary']=df['monthly_salary'].astype(str)
df['monthly_salary']=df['monthly_salary'].str.replace(r'\.0+', '', regex=True)
df['monthly_salary']=pd.to_numeric(df['monthly_salary'])
```

### Fxing Monthly_rent column imputed it with median(0)

```python
df['monthly_rent']=df['monthly_rent'].fillna(df['monthly_rent'].median())
```

### removed anomalies and nulls from credit_score

```python
df=df[
    (df['credit_score'].between(300,850))&
    (df['credit_score'].notnull())
]
```

```python
df.shape
```

-- standardizing bank_balance

```python
df['bank_balance']=df['bank_balance'].astype(str)
df['bank_balance']=df['bank_balance'].str.replace(r'\.0+', '', regex=True)
df['bank_balance']=pd.to_numeric(df['bank_balance'],errors='coerce')
```

```python
df['bank_balance'].isnull().sum()
```

### as bank balance is strongly related to our target so we are dropping rows 

```python
df=df[df['bank_balance'].notnull()]
```

```python
df=df[df['emergency_fund'].notnull()]
```
